```python
### The if Construct
# The sample code below shows how the basic if construct
# works in Python.
value = False

print("Hello")

# value below is a Boolean variable. The indented code
# lines below the if line will execute only when the
# Boolean is True. That code is skipped otherwise.
if value:
  print("How are you?")

# The indenting is now removed, so the code below runs
# regardless of the Boolean value. It is NOT guarded
# by an if conditional.
print("Good bye")


```

```python
### if - else constructs
# The if clause can also have at most one associated
# else clause. The else clause must come after the
# paired if clause, and you cannot have any other
# code separating the if and else clauses.

value2 = False
print("Hello")

# value2 below is a Boolean variable. If value2 is True,
# the indented code below the if clause will execute
# and the code indented under the else clause will be
# skipped. If value2 is False, the opposite occurs.
if value2:
  print("How are you?")
else:
  print("I hope you are well!")

# The if-else clauses are now over. The code below
# is not indented, so it will run regardless of the value
# in value2.
print("Good bye")
```
